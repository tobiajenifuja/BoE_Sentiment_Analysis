```python
import nbformat
from nbformat import NotebookNode

def merge_notebooks(filenames):
    merged = nbformat.v4.new_notebook()
    merged.cells = []

    for fname in filenames:
        nb = nbformat.read(fname, as_version=4)
        merged.cells.extend(nb.cells)

    return merged

files = [
    "../master_notebook_files/(UK) Vader Sentiment Analysis with LM Dictionary.ipynb",
    "../master_notebook_files/(Euro) Vader Sentiment Analysis with LM Dictionary.ipynb",
    "../master_notebook_files/(US) Vader Sentiment Analysis with LM Dictionary.ipynb",
    "../master_notebook_files/(Canada) Vader Sentiment Analysis with LM Dictionary.ipynb",
    "../master_notebook_files/(Japan) Vader Sentiment Analysis with LM Dictionary.ipynb"
]

merged = merge_notebooks(files)
nbformat.write(merged, "insightnexus_merged_notebook_test.ipynb")
```
