```python
import nbformat
from nbformat import NotebookNode

def merge_notebooks(filenames):
    merged = nbformat.v4.new_notebook()
    merged.cells = []

    for fname in filenames:
        nb = nbformat.read(fname, as_version=4)
        merged.cells.extend(nb.cells)

    return merged

files = [
    "../raw_notebooks/insightnexus_final_notebook.ipynb",
    "../raw_notebooks/merged_all_vader_sentiment_analysis_with_lm_dictionary.ipynb",
    "../raw_notebooks/Full - FinBERT Sentiment Analysis - BoE Speeches.ipynb",
    "../raw_notebooks/FULL Regression Analysis (Sentiment vs Key variables)v2.ipynb",
    "../raw_notebooks/BoE_FinancialStability_SentimentAnalysis_v.2.ipynb",
    "../raw_notebooks/BoE_IR-MPR_Sentiment_Analysis_LM_VADER_v1.ipynb",
    "../raw_notebooks/mpc_voting.ipynb",
    "../raw_notebooks/LDA - Master Notebook.ipynb"
]

merged = merge_notebooks(files)
nbformat.write(merged, "insightnexus_final_notebook_draft_v1.ipynb")
```
